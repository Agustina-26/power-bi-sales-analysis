```python
import pandas as pd
```

```python
df = pd.read_csv('ventasTransformed.csv')
```

```python
print(df.columns.tolist())
print(df.shape)

df.head()
```

```python
clientes = (df[[
        'ClienteNombre',
        'ClienteGenero',
        'ClienteEdad',
        'ClienteMail',
        'ClienteTelefono',
        'ClienteDireccion']
].drop_duplicates().reset_index(drop=True))    
clientes['ClienteID'] = clientes.index + 1
clientes = clientes[['ClienteID', 'ClienteNombre', 'ClienteGenero', 'ClienteEdad', 'ClienteMail', 'ClienteTelefono', 'ClienteDireccion']]

vendedores = (df[['VendedorNombre']]).drop_duplicates().reset_index(drop=True)
vendedores['VendedorID'] = vendedores.index + 1
vendedores = vendedores[['VendedorID', 'VendedorNombre']]

ciudades = df[
        'SucursalCiudad'
].unique()
ciudades = pd.DataFrame(ciudades, columns=['Ciudad'])
ciudades['CiudadID'] = ciudades.index + 1
ciudades = ciudades[['CiudadID', 'Ciudad']]

sucursales = (df[[
        'SucursalNombre',
        'SucursalCiudad'
]].drop_duplicates().reset_index(drop=True))
sucursales['SucursalID'] = sucursales.index + 1
sucursales = sucursales.merge(
    ciudades,
    left_on = 'SucursalCiudad',
    right_on = 'Ciudad',
    how = 'left'
)
sucursales = sucursales[['SucursalID', 'SucursalNombre', 'SucursalCiudad', 'CiudadID']]

MetodoPago = df['MetodoPago'].dropna().unique()
MetodoPago = pd.DataFrame(MetodoPago, columns=['MetodoPago'])
MetodoPago ['MetodoPagoID'] = MetodoPago.index + 1
MetodoPago = MetodoPago[['MetodoPagoID', 'MetodoPago']]
```

```python
sucursales[sucursales['SucursalNombre'].duplicated(keep=False)]
```

### Tabla Productos


```python
productos = pd.DataFrame(columns = ['NombreProducto', 'MarcaProducto', 'Precio'])

for x in range(1, 4): 
    append_producto = pd.DataFrame(columns = ['NombreProducto', 'MarcaProducto', 'Precio'])
    append_producto['NombreProducto'] = df[f'NombreProducto{x}']
    append_producto['MarcaProducto'] = df[f'MarcaProducto{x}']
    append_producto['Precio'] = df[f'PrecioUnitarioProducto{x}']
    productos = pd.concat([productos, append_producto], ignore_index=True)


productos = productos.dropna(subset=['NombreProducto','MarcaProducto','Precio'])
productos = productos.drop_duplicates(['NombreProducto','MarcaProducto'])
productos['ProductoID'] = productos.index + 1
productos = productos[['ProductoID', 'NombreProducto', 'MarcaProducto', 'Precio']]
```

### Para crear la tabla de Facturas, debemos hacer un merge usando los ID que creamos hasta ahora.

```python
facturas = df.copy()
facturas = facturas.groupby('VentaID', as_index=False).first()

facturas = facturas[[
    'VentaID',
    'ClienteNombre',
    'ClienteEdad',
    'ClienteGenero',
    'ClienteMail',
    'SucursalNombre',
    'SucursalCiudad',
    'VendedorNombre',
    'MetodoPago',
    'DescuentoVenta',
    'TotalVenta',
    'Año',
    'MesNombre',
    'Día'
]]

facturas = facturas.merge(
    clientes[['ClienteID', 'ClienteNombre', 'ClienteMail']],
    on=['ClienteNombre', 'ClienteMail'],
    how='left'
)

facturas = facturas.merge(
    vendedores[['VendedorID', 'VendedorNombre']],
    on='VendedorNombre',
    how='left'
)

facturas = facturas.merge(
    sucursales[['SucursalID', 'SucursalNombre', 'SucursalCiudad']],
    on=['SucursalNombre', 'SucursalCiudad'],
    how='left'
)
```

### Tabla Detalles

```python
DetalleFacturas = pd.DataFrame(columns = ['VentaID', 'ProductoID', 'Cantidad', 'Subtotal'])

for i in range(1, 4):
    detalle_temp = df[['VentaID', f'NombreProducto{i}', f'MarcaProducto{i}', f'CantidadProducto{i}', f'SubtotalProducto{i}']]
    detalle_temp = detalle_temp.merge(productos, left_on=[f'NombreProducto{i}', f'MarcaProducto{i}'], right_on=['NombreProducto', 'MarcaProducto'], how='left')
    detalle_temp = detalle_temp[['VentaID', 'ProductoID', f'CantidadProducto{i}', f'SubtotalProducto{i}']]
    detalle_temp = detalle_temp.rename(columns={f'CantidadProducto{i}': 'Cantidad', f'SubtotalProducto{i}': 'Subtotal'})

    DetalleFacturas = pd.concat([DetalleFacturas, detalle_temp], ignore_index=True)

DetalleFacturas = DetalleFacturas.drop_duplicates(['VentaID','ProductoID'])


DetalleFacturas = DetalleFacturas.dropna().reset_index(drop=True)
DetalleFacturas['Cantidad'] = DetalleFacturas['Cantidad'].astype(int)
DetalleFacturas['Subtotal'] = DetalleFacturas['Subtotal'].astype(float)
DetalleFacturas['VentaID'] = DetalleFacturas['VentaID'].astype(int)
DetalleFacturas['ProductoID'] = DetalleFacturas['ProductoID'].astype(int)

DetalleFacturas.sort_values(by=['VentaID'], inplace=True)
DetalleFacturas.reset_index(drop=True, inplace=True)

#Creamos un ID para cada detalle

DetalleFacturas['DetalleID'] = DetalleFacturas.index + 1
DetalleFacturas = DetalleFacturas[['DetalleID', 'VentaID', 'ProductoID', 'Cantidad', 'Subtotal']]
```

## Exportamos a un archivo Excel 

```python
dataframes = [
    ciudades,
    sucursales,
    vendedores,
    clientes, 
    MetodoPago, 
    productos,
    facturas,
    DetalleFacturas
]

nombres_hojas = [
    'Ciudades', 
    'Sucursales', 
    'Vendedores',
    'Clientes',
    'MetodosPago',
    'Productos',
    'Facturas',
    'DetalleFacturas'
]

ruta_del_archivo = 'modeloVentas.xlsx'

with pd.ExcelWriter(ruta_del_archivo, engine='xlsxwriter') as writer:
    for nombre_hoja, df in zip(nombres_hojas, dataframes):
        df_export = df.copy()
        df_export = df_export.fillna({
            'ClienteNombre': '',
            'VendedorNombre': '',
            'SucursalNombre': ''
        })
        df_export.to_excel(writer, sheet_name=nombre_hoja, index=False)
```
